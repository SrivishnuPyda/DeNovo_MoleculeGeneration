# Random Notebook 5
This notebook explores matrix operations using NumPy.

## Matrix Example 1
Matrix manipulations with NumPy.

```
# Matrix operations Example 1
import numpy as np

A = np.random.randint(1, 10, size=(3, 3))
B = np.random.randint(1, 10, size=(3, 3))

C = np.dot(A, B)
detA = np.linalg.det(A)

print("Example 1:")
print("Matrix A:\n", A)
print("Matrix B:\n", B)
print("A*B:\n", C)
print("det(A):", detA)
```

In Example 1, we multiplied two random matrices and computed the determinant of A.
