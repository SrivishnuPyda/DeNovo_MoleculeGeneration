# Random Notebook 3
This notebook explores basic statistics and random generation.

## Part 1
Exploring random statistics with Python.

```
# Random statistics Part 1
import statistics, random

samples = [random.gauss(1*2, 1+1) for _ in range(50)]
mean_val = statistics.mean(samples)
median_val = statistics.median(samples)
stdev_val = statistics.stdev(samples)

print("Part 1: Mean =", mean_val, "Median =", median_val, "StdDev =", stdev_val)
```

In Part 1, we generated Gaussian samples and computed basic statistics.
